# I. Introduction

Every second counts; the decisions made in split seconds by players and coaches significantly impact the outcome of a play. One critical aspect of modern football strategy is predicting the likelihood of a caught forward pass, especially within the context of rapidly changing plays, schemes, formations and player alignment.

This paper presents a **novel approach to predicting the probability of a caught
forward pass within the next 0.5 seconds given the current play configuration and the correspoding contextual information**: pre-snap configurations (such as line-set and ball-snap positions) and game contextual information. **This is done by leveraging tracking data as a graph-based modeling technique.**

Understanding these patterns can improve both offensive planning and defensive strategies, providing a competitive edge on game day.

Anticipating forward passes requires a detailed understanding of individual player movements, team formations, and the dynamic interactions between opposing teams. NFL tracking data serves as a vital resource for such analysis, capturing the positions and movements (referred to as configurations) of all 22 players and the ball on the field at a high temporal resolution of 10 frames per second (fps).

Previous research in football analytics has utilized traditional machine learning and statistical methods for predictive modeling. However, these approaches often fall short in capturing the intricate complexities of player interactions. A graph-based representation addresses this limitation effectively by modeling the dynamics of the play more naturally. In this representation, nodes correspond to individual players and the ball, while edges encode the interactions among them. This graph framework offers a holistic view of each frame’s configuration, facilitating deeper insights into the strategies and outcomes of plays.

# I. Methodology

### a. Graph Representation
A graph is used to model the interactions of key components on the field, namely the players and the ball. These components interact in various ways and have distinct attributes, which are represented through specific physical characteristics. We employ a Multidigraph data structure to represent these interactions, as it allows for directed edges and multiple connections between pairs of nodes, capturing the meaning behind individual players interacting with other players and the strenght of these interactions.

<img src="https://raw.githubusercontent.com/dhdzmota/NFL-Big-Data-Bowl-2025/refs/heads/from_local_to_remote/reports/report-imgs/graph-representation.png" width="500"/>
</center>



#### i. Node Attributes:
Nodes represent individual players and the ball, each characterized by the following attributes:
  - Action: $msd$
  - Kinetic Energy: $\frac{1}{2}ms^2$
  - Momentum: $ms$
  - Force: $ma$
  - Work: $mad$
  - Team redeffinition: -1 for the defensive team, 1 for the possesion team.
  - Player position: A numerical representation of the players role.
  - Player location: x and y coordinates of the player in the field.

Where, for each player:
  - $m$ is the weight (according to the weight in "player data" dataset)
  - $s$ is the speed in yards/second
  - $a$ is the acceleration in yards/second$^2$,
  - $d$ is the distance traveled from prior time point in yards.

#### ii. Edges:

Edges represent the interaction between players as the proximity (p), computed as a function of distance $p(distance) = (1/distance)$,  weighted by a metric we term _look_, which measures motion direction and orientation alignment. The _look_ metric is defined as:

$$look = \frac{1}{2}(look_{ω} + look_{δ})$$

To compute each component, a variable $(α)$ is defined as the angle generated from the straight line connecting two players positions in the field. Then, the difference between the players orientation $(ω)$ and direction $(δ)$ angles are calculated.

- Orientation contribution $look_{ω} = cos(\alpha-ω)$: When the angle difference $\alpha-ω$ approaches 0, the cosine value approaches 1, indicating alignment between the field of view and the other players.

- Direction Contribution $look_{δ} = cos(\alpha-δ)$ :
This component follows the same logic, capturing alignment in directional movement.

In these instances, negative values of look are not considered as an interaction, since alignment is in an oposite direction.

<center>
<img src="https://raw.githubusercontent.com/dhdzmota/NFL-Big-Data-Bowl-2025/refs/heads/from_local_to_remote/reports/report-imgs/node-interaction-representation.png" width="500"/>
</center>

Therefore an edge, or interaction $(I)$, is computed as the following formula:

$$I = look*p(distance)$$

### b. Modeling

To model, we employ traditional machine learning algorithms alongside graph embeddings to capture the essence of each play snapshot.

The training procedure involved analyzing each frame to determine the event occurring in the next 0.5 seconds. If a forward pass was made and successfully caught, we assigned a True value; otherwise, a False value. This created a dichotomous response variable. Relevant events were selected to train the model to distinguish when a configuration would yield the expected result and when it would not. Additionally, random configurations where nothing relevant happened were also included so that the model would also use that knowledge to leverage the predictions.

#### i. Input Features

The input features are divided into broad categories:

- Initial Features: These correspond to the play context, including: Quarter, Down, Yards to Go, Yardline Number, Pre-snap Score, Win Probability Difference between the two teams, Play Clock, Player Alignments.

- Mechanical Features: These capture the mechanical behavior of the players as a whole, including: Kinetic Energy, Force, Momentum, Work, Action, Team Location, centroids. Then we made team comparisons (possession vs. defensive team).

- Field Influence: A composite value that combines player movement attributes .

- Distance Metrics: Features measuring distances between teams and tracking the football’s mechanical characteristics.

- Event Features: A sequential list of player positions at the events before the snap.

- Graph Embeddings: Using the Feather technique (Rozemberkczki & Sarkar, 2020), we incorporate graph topology and specific node attributes into the model. To enhance the model performance we applied dimensionality reduction method (PCA) to these embeddings.

#### ii. Output

The model outputs a probability score indicating the likelihood of a caught forward pass exactly 0.5 seconds into the future.

This graph-based approach captures dynamic player interactions and integrates them into a predictive framework, providing actionable insights into play outcomes.

It is important to note that features are computed at a frame-level resolution and are analyzed exclusively before the snap. As a result, the model enables post-snap evaluation, generating an evaluation window to identify different opportunities for executing a forward pass that is likely to be caught.

# II. Results

## a. Model Performance

The model demonstrated promising performance. By choosing the best threshold evaluating each possible cutpoint,   It correctly identified 89% of samples that do not belong to the target positive class and 86% of samples that do. However, the model exhibits some over-enthusiasm, with a precision of 59%, indicating that it made a wrong decision of tagging the positive class 41% of the time. These metrics result in an F1-score of 0.7, indicating that while the model provides trustworthy predictions, there is room for further optimization to enhance its precision.

## b. Evaluation and Insights

By analyzing an entire play after the snap, the model generates a probability score for each frame. This creates a decision window that allows for in-depth analysis to identify the optimal moment to execute a forward pass. The analysis also highlights when this window closes, thereby reducing the likelihood of a successful forward pass.

<center>
<img src="https://raw.githubusercontent.com/dhdzmota/NFL-Big-Data-Bowl-2025/refs/heads/from_local_to_remote/reports/report-imgs/insights.gif" width="500"/>
</center>

To further understand model behavior, we employed feature importance analysis using SHAP values (Lundberg & Lee, 2017) to identify the most influential features. SHAP values were calculated for each frame across a game or a specific time window. By aggregating these values across frames, we determined the top 10 features influencing predictions. Notably, 9 out of the 10 most important features were derived from graph embeddings, underscoring the significance of the graph-based approach in capturing critical interactions.


<center>
<img src="https://raw.githubusercontent.com/dhdzmota/NFL-Big-Data-Bowl-2025/refs/heads/from_local_to_remote/reports/report-imgs/shap-features.png" width="700"/>
</center>

In the figure above, each feature's behaviour over time is illustrated. The values are scaled using a min-max normalization function within the range [0, 1], with the mid-point represented by a dotted line. Each frame is represented as a point, and the corresponding SHAP values are color-coded: positive values indicate contributions to the positive class, while negative values indicate contributions to the negative class. This visualization highlights how feature contributions fluctuate over time and their relative impact on model predictions. Furthermore, the features are ordered by importance, with the top feature being the most impactful (rank 1), followed by subsequent features in descending order of influence.

# Conclusions

This study highlights the transformative potential of graph-based analysis in football strategy, offering a robust framework for evaluating distinct configurations and predicting forward pass outcomes. By modeling players and their interactions as graph components, we effectively capture the intricate dynamics of on-field play, bridging traditional football analysis with advanced data science methodologies.

From a football perspective, this approach delivers actionable insights for coaches and analysts, enabling the identification of optimal decision windows and the refinement of both offensive and defensive strategies. On offense, critical plays—those reserved for pivotal moments such as fourth downs or game-altering scenarios—can be revolutionized with predictive insights, especially when tailored to specific opponents. On defense, this framework offers a deeper understanding of opposing strategies, facilitating the anticipation of key plays and the identification of real-time vulnerabilities. By customizing defensive formations and assignments based on predictive outcomes, teams can disrupt high-probability plays and compel less favorable decisions from their rivals. Additionally, the model's capacity to analyze post-snap configurations provides a valuable resource for strategic planning, in-game decision-making, and performance evaluations.

From a data science standpoint, this work underscores the power of integrating advanced machine learning techniques with domain-specific expertise. The use of graph-based analysis illustrates how data-driven methodologies can yield granular, actionable insights in highly complex and dynamic contexts like football.

In conclusion, this graph-based framework marks a significant advancement in football analytics. It not only enhances strategic planning and real-time decision-making but also demonstrates the potential of merging cutting-edge data science with the art and strategy of football, shaping the future of the game.
